# 6 Waves on a Beam
## 6.1 Wave Types
![](tikzfig/beam_cross_section.svg)
* cross-section area $S$
* stress-free at the whole circumference

### Purely axial load (bar)
* assume "thin" cross-section
* transverse normal stresses vanish $\sigma_{y}=0, \sigma_{z}=0$
* normal stress for purely axial load 
\begin{equation}\label{eq:stressLbar}
\sigma_{x}={2 G}(1+\mu)\frac{\partial \xi_x}{\partial x}=E\frac{\partial \xi_x}{\partial x}
\end{equation}
* transverse contraction is unhindered
* shear stress does not vary over cross section
* insert into equation of motion, this gives
\begin{equation}\label{eq:Lwavebeam}
\frac{\partial^2 \xi_x}{\partial x^2} -\frac{\rho}{E}\frac{\partial^2 \xi_x}{\partial t^2}=0
\end{equation}
*  wave equation for quasilongitudinal waves with speed somewhat lower than for those in plates
\begin{equation}\label{eq:cLII}
c_{LII}=\sqrt{\frac{E}{\rho}}
\end{equation}

### Purely torsional load (shaft/beam)
![](tikzfig/beam_torsion.svg)
* torsion around $x$-axis with angle $\chi_x$
* after integration of shear stresses over cross section we get
\begin{equation}\label{eq:tmoment}
M_x =G J_{T}\frac{\partial \chi_x}{\partial x}=T\frac{\partial \chi_x}{\partial x}
\end{equation}
where $M_x$ is the torsional moment and $T$ is the torsional stiffness
* torsion constant $J_T$ around shear center is a second moment, depends on the cross sectional shape (see handbooks for tabluated values)
* _only_ for circular shape with radius $R$:
\begin{equation}\label{eq:JTcirc}
J_{T}=\frac{\pi}{2}R^4
\end{equation}
* rotational inertia
\begin{equation}\label{eq:}
M_x = \Theta'_x \frac{\partial^2 \chi_x}{\partial t^2}
\end{equation}
with the moment of inertia $\Theta'_x = \rho I_p$ and $I_p=\int_0^R r^2 dr$ the polar moment of inertia with $r$ beeing the distance to mass center
* for circular and rectangular shape ($h\times b$)
\begin{equation}\label{eq:Ip}
I_{p}=\frac{\pi}{2}R^4,\:\:\:\: I_{p}=\frac{bh^3+hb^3}{12}
\end{equation}
* from equation of motion:
\begin{equation}\label{eq:Twavebeam}
\frac{\partial^2 \chi_x}{\partial x^2} -\frac{\rho I_p}{GJ_x}\frac{\partial^2 \chi_x}{\partial t^2}=0
\end{equation}
*  wave equation for torsional waves with speed equal or slower than transversal waves ($c_T$)
\begin{equation}\label{eq:cTI}
c_{TI}=\sqrt{\frac{T}{\Theta'_x}}=c_T\sqrt{\frac{J_T}{I_p}}
\end{equation}

### Purely bending load (beam)
![](tikzfig/beam_bending.svg)
* no shear: relation between rotation angle around the $z$-axis and transverse displacement is
\begin{equation}\label{eq:bend1}
\chi_z=\frac{\partial \xi_y}{\partial x}
\end{equation}
* axial stress if integrated over the cross section
* bending moment relates to curvature:
\begin{equation}\label{eq:bend2}
M_z=-B_z\frac{\partial \chi_z}{\partial x}
\end{equation}
with bending stiffness $B_z=EI_z=E\int_{S}y^2dS$
* transversal force relates to change of moment
\begin{equation}\label{eq:bend3}
F_y= -\frac{\partial M_z}{\partial x}
\end{equation}
* change of force results in transverse acceleration
\begin{equation}\label{eq:bend4}
-\frac{\partial F_y}{\partial x}= m'\frac{\partial^2 \xi_y}{\partial t^2}
\end{equation}
with mass per unit length $m'=\rho S$

```python
import sympy as sy
x, t = sy.symbols("x t")
B, m1 = sy.symbols("B m'", positive=True)
Mz = sy.Function('M_z')(x,t)
Fy = sy.Function('F_y')(x,t)
chiz = sy.Function('chi_z')(x,t)
xiy = sy.Function('xi_y')(x,t)
s1 = {chiz:sy.diff(xiy,x)}
s2 = {Mz:-B*sy.diff(chiz,x)}
s3 = {Fy:-sy.diff(Mz,x)}
eq4 = sy.Eq(-sy.diff(Fy,x),m1*sy.diff(xiy,t,2))
eq4.subs(s3).subs(s2).subs(s1).doit()
```

* fourth order PDE - bending wave equation
* similar to plate:
\begin{gather}\label{eq:bendingkc}
k_B^4=\frac{m'\omega^2}{B}\\
c_B=\frac{\omega}{k_B}=\pm\sqrt{\omega}\sqrt[4]{\frac{B}{m'}}\\
c_{Bg} = 2 c_B
\end{gather}
* a _second_ bending wave travels along the beam with displacements mainly in $z$-direction

## 6.2 Energy Transport
* Intensity (rate of energy transport)
\begin{equation}\label{eq:energy_transport}
I = c_g w
\end{equation}
* bending wave energy density (four components)
  1. potential energy due to axial stresses
  2. potential energy due to transversal stresses (negligibly small)
  3. kinetic energy due to transversal displacement
  4. kinetic energy due to rotation of beam section around $z$-axis (negligibly small)
\begin{gather}\label{eq:wpot}
S w_{pot}=\frac{B}{2}\left(\frac{\partial^2 \xi_y}{\partial x^2}\right)^2\\
\label{eq:wkin}
S w_{kin}=\frac{m'}{2}\left(\frac{\partial \xi_y}{\partial t}\right)^2
\end{gather}
* wave traveling in positive $x$ direction 
\begin{equation}
\xi_y=\xi_{y0} \cos(\omega t- k_B x)
\end{equation}
* calculate power of the wave (energy per unit time):
\begin{equation}\label{eq:powerb}
P = I S
\end{equation}

```python
om = sy.symbols('omega',positive=True)
xiy0 = sy.symbols('xi_y0')
kB = sy.sqrt(om) * sy.root(m1/B,4) # wave number
cg = 2*om/kB # group speed
xiy_ = xiy0*sy.cos(om*t-kB*x) # wave in pos. x direction
# energy per unit length S*w 
Swpot = B/2 * sy.diff(xiy,x,2)**2
Swkin = m1/2 * sy.diff(xiy,t)**2
# substitute the wave
wp = Swpot.subs(xiy,xiy_).doit()
wk = Swkin.subs(xiy,xiy_).doit()
# display energy densities
display(wp,wk)
# power
P = sy.simplify(cg*(wk+wp)).doit()
# display the power
P
```

* mechanical power can also be derived from work done per unit time:
  * from force and
  \begin{equation}\label{eq:power1}
  P_1=F_y v_y
  \end{equation}
  * from moment
    \begin{equation}\label{eq:power2}
    P_2=M_z w_z
    \end{equation}
    where $w_z=\frac{\partial \chi_z}{\partial t}$ is rotational speed

```python
# two components of power
P1 = Fy*sy.diff(xiy,t) # vy from xiy
P2 = Mz*sy.diff(chiz,t) # wz from chiz
# substitute to have only xiy dependency
P1_ = P1.subs(s3).subs(s2).subs(s1).subs(xiy,xiy_).doit().simplify()
P2_ = P2.subs(s3).subs(s2).subs(s1).subs(xiy,xiy_).doit().simplify()
# overall power
P_ = sy.simplify(P1_+P2_)
display(P_)
# compare to other result for power
sy.simplify(sy.Eq(P,P_))
```

* as expected both results are the same

## 6.3 Excitation of Longitudinal and Bending waves
![](tikzfig/beam_end.svg)

* free condition at end: forces and moments are prescribed, displacements and rotations are free
* excitation with a harmonic 'point' force acting at end

**Longitudinal wave excitation at end ('half-infinite' beam)**
* axial force produces quasilongitudinal wave in +$x$ direction
* given: $F_x$ at end $x=0$
\begin{equation}\label{eq:Fx}
F_x=-\int_S \sigma_{xx}dS=-S\sigma_{xx}=-SE\frac{\partial \xi_x}{\partial x}
\end{equation}
* determine impedance at end
\begin{equation}\label{eq:ZLII}
\Zc_{LII}=\left.\frac{\Fc}{\vc}\right|_{x=0}
\end{equation}


```python
# constants
S, E, rho = sy.symbols('S E rho',positive=True)
# wave amplitude
xix0 = sy.symbols('xi_x0')
# qlongitudinal wave speed and wave number
cLII = sy.sqrt(E/rho)
kL = om/cLII
# wave travelling into pos. x direction
xix_ = xix0*sy.exp(-sy.I*kL*x)
# Fx and vx
Fx = -S*E*sy.diff(xix_,x)
vx = sy.I*om*xix_
# impedance_
ZLII = sy.simplify((Fx/vx).subs(x,0))
ZLII
```

\begin{equation}
Z_{LII}=\sqrt{E \rho}S=m'c_{LII}
\end{equation}


**Bending wave excitation at end ('half-infinite' beam)**
* transversal force $F_y$ produces two bending waves in +$x$ direction at the same time:
  1. traveling wave
  2. evanescent "wave" ("near field")
* both amplitudes are unknown
* from boundary conditions: $M_z=0$
* determine impedance at end:

```python
# evanescent wave amplitude
xiy1 = sy.symbols('xi_y1')
# bending wave speed
cB = sy.symbols('c_B', positive=True)
# traveling wave and propagation wave - two amplitudes
xiy_ = xiy0*sy.exp(-sy.I*kB*x)+ xiy1*sy.exp(-kB*x)
# set up Mz in terms of xiy and solve for evanescent amplitude
Mz_ = Mz.subs(s2).subs(s1).subs(xiy,xiy_).doit()
xiy1_ = sy.solve(Mz_.subs(x,0),xiy1)[0]
# show result for evanescent amplitude
display(sy.Eq(xiy1,xiy1_)) # result: both amplitudes are the same!
# substitute evanescent amplitude
xiy__ = xiy_.subs(xiy1,xiy1_)
# Fy and vy
Fy_ = Fy.subs(s3).subs(s2).subs(s1).subs(xiy,xiy__).doit()
vy = sy.I*om*xiy__ # v = d xi / d t
# impedance
ZB = sy.simplify((Fy_/vy).subs(x,0))
display(ZB)
# rewrite using the bending wave speed
B_ = sy.solve(sy.Eq(cB,om/kB),B)[0]
ZB.subs(B,B_).simplify()
```

* impedance is complex-valued
* imaginary and real part are the same
* 45° phase shift between force and velocity
* amplitudes for both traveling and evanescent wave are the same
* let us plot the displacement of the beam and the force at the end:

```python
import numpy as np
import matplotlib.pyplot as pl
import ipywidgets as ipw

@ipw.interact(t=(0,10,0.1))
def plot_xiy_halfbeam(t=0):
    x = np.arange(0,15,0.1)
    # assume amplitudes=1, kB=1, omega=1
    xi = (np.exp(-1j*x)+np.exp(-1*x))*np.exp(1j*t)
    # assume m'=1, 
    # F = Z v = Z j omega xi:
    F = (1+1j)/2*1j*xi[0]
    # plot everything
    pl.plot(x,xi.real,label=r'$\xi_y$')
    pl.plot(0,F.real,'ro',label=r'$F(x=0)$')
    pl.ylim(-3,3)
    pl.xlabel(r'$x$')
    pl.legend()
```

* evanescent wave contributes only near the excitation

**Bending wave excitation of an infinite beam**

![](tikzfig/beam_mid.svg)
* transversal force at $x=0$ produces four waves:
  1. traveling wave in +$x$ direction
  2. evanescent "wave" ("near field") in +$x$ direction
  1. traveling wave in -$x$ direction
  2. evanescent "wave" ("near field") in -$x$ direction
* problem is symmetric - we treat only the first two waves, the amplitudes of the latter must be the same!
* boundary condition for excitation at $x=0$ is not free, instead $\chi_z=0$ and $\xi_y$ and $M_z$ are free
* determine impedance:

```python
# set up wz in terms of xiy and solve for evanescent amplitude
chiz_ = chiz.subs(s1).subs(xiy,xiy_).doit().simplify()
xiy1_ = sy.solve(chiz_.subs(x,0),xiy1)[0]
# show result for evanescent amplitude
display(sy.Eq(xiy1,xiy1_))
# substitute evanescent amplitude
xiy__ = xiy_.subs(xiy1,xiy1_)
# Fy and vy
Fy_ = 2*Fy.subs(s3).subs(s2).subs(s1).subs(xiy,xiy__).doit()
vy = sy.I*om*xiy__
# impedance
ZB = sy.simplify((Fy_/vy).subs(x,0))
display(ZB)
# rewrite using the bending wave speed
ZB.subs(B,B_)
```

* equal to four times the impedance of the half-infinite beam
* amplitude of the evanescent wave has the same magnitude but a 45° phase shift:

```python
@ipw.interact(t=(0,10,0.1))
def plot_xiy_fullbeam(t=0):
    x = np.arange(-15,15,0.1)
    # assume amplitudes=1, kB=1, omega=1
    xi = (np.exp(-1j*np.abs(x))-1j*np.exp(-1*np.abs(x)))*np.exp(1j*t)
    # assume m'=1, 
    # F = Z v = Z j omega xi:
    F = 2*(1+1j)*1j*xi[0]
    # plot everything
    pl.plot(x,xi.real,label=r'$\xi_y$')
    pl.plot(0,F.real,'ro',label=r'$F(x=0)$')
    pl.ylim(-3,3)
    pl.xlabel(r'$x$')    
    pl.legend()
```

## 6.4 Input Impedance and Power Input
* input of power for a given excitation
* from [](#eq:power1) and [](#eq:power2):
\begin{equation}\label{eq:poweri1}
P = P_1+P_2= Fv+Mw
\end{equation}
* if $M=0$ or $w=0$ (true for all cases above):
\begin{equation}\label{eq:poweri2}
\overline{P} = \overline{F(t)v(t)} = \overline{\Re(\Fc) \Re(\vc)}=\frac{1}{2}\Re(\Fc\vc^*)
\end{equation}
* power input is then always
\begin{equation}\label{eq:poweriZ}
P = \frac{1}{2}|\Fc|^2\Re(\frac{1}{\Zc})
\end{equation}
* depending on input impedance, power may be frequency dependent
* example for bending wave impedance of infinite beam:
\begin{equation}\label{eq:poweribend}
P= \frac{|\Fc|^2}{\omega m' c_B}
\end{equation}
